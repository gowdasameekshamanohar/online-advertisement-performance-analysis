Which platform generated the most clicks?

```python
plt.bar(df["platform"], df["clicks"])
plt.title("clicks in each platform")
plt.xlabel("platform")
plt.ylabel("No. of clicks")
plt.show()
```

```python
clicks_by_platform = df[["platform", "clicks"]].sort_values(
    by="clicks",
    ascending=False
)
clicks_by_platform
```

which platform is better at turning views(impressions) into clicks? 
(using Click-Through-Rates or CTR)

```python
plt.bar(df["platform"], df["ctr"])
plt.title("impressions that turned into clicks")
plt.xlabel("platform")
plt.ylabel("CTR (%)")
plt.show()
```

```python
df[["platform", "ctr"]].sort_values(
    by="ctr",
    ascending=False
)
```

```python
which platform has most conversions? 
```

```python
plt.bar(df["platform"], df["conversions"])
plt.title("platforms with no. of convertions")
plt.xlabel("platform")
plt.ylabel("no. of conversions")
plt.show()
```

```python
df[["platform", "conversions"]].sort_values(
    by="conversions",
    ascending=False
)
```

percentage of users who clicked and actually converted?

```python
plt.bar(df["platform"], df["conversion_rate"])
plt.title("Clicked users who got converted")
plt.xlabel("platform")
plt.ylabel("conversion rate (%)")
plt.show()
```

```python
df[["platform", "conversion_rate"]].sort_values(
    by="conversion_rate",
    ascending=False
)
```

how much did the advertiser spend on each click on average? 
(Cost-Per-Click or CPC)

```python
plt.bar(df["platform"], df["avg_cpc"])
plt.title("average spending per click on each platform")
plt.xlabel("platform")
plt.ylabel("cpc (in $)")
plt.show()
```

```python
df[["platform", "avg_cpc"]].sort_values(
    by="avg_cpc",
    ascending=True #using the value as true because the lower the value, the more cost efficiency
)
```

how expensive it is to reach 1,000 impressions

```python
plt.bar(df["platform"], df["cpm"])
plt.title("cost on each platform to reach 1000 views")
plt.xlabel("platform")
plt.ylabel("cpm (in $)")
plt.show()
```

```python
df[["platform", "cpm"]].sort_values(by="cpm", ascending=True)
```

does spending more money actually lead to more conversions?

```python
plt.scatter(df["cost"], df["conversions"])
plt.title("spending vs conversions")
plt.xlabel("Ad spend (in $)")
plt.ylabel("conversions")
plt.show()
```

how much money was spent to get one conversion?

```python
df["cost_per_conversion"] = df["cost"] / df["conversions"]
```

```python
df[["platform", "cost", "conversions", "cost_per_conversion"]]
```

```python
plt.bar(df["platform"], df["cost_per_conversion"])
plt.title("money spent on getting one conversion")
plt.xlabel("platform")
plt.ylabel("cost per conversion (in $)")
plt.show()
```

```python
df[["platform", "cost_per_conversion"]].sort_values(
    by="cost_per_conversion",
    ascending=True
)
```
